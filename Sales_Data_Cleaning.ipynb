```python
# Import libraries
import pandas as pd
import numpy as np
```

```python
#Load the dataset
df=pd.read_csv('Messy_data.csv')
```

```python
# Remove extra spaces from column names
df.columns=df.columns.str.strip()

# Check column names
print(df.columns.tolist())

# Remove duplicate column names, keeping the first
df=df.loc[:,~df.columns.duplicated()]
```

```python
df=df.drop_duplicates()
```

```python
print(df.duplicated().sum())
```

```python
df.shape
```

```python
#Clean Extra spaces and standardize the text
text_columns=["CustomerName","Product","City","OrderStatus"]
df["CustomerName"]=df["CustomerName"].str.title()
df["City"]=df["City"].str.title()
df["Product"]=df["Product"].str.title()
df["OrderStatus"]=df["OrderStatus"].str.title()
```

```python
for col in text_columns:
  df[col]=df[col].astype("string").str.strip()
```

```python
#Outliers-Price
df["UnitPrice"]=pd.to_numeric(df["UnitPrice"],errors="coerce")
Q1=df["UnitPrice"].quantile(0.25)
Q3=df["UnitPrice"].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
```

```python
df['UnitPrice']=df["UnitPrice"].clip(lower,upper)
```

```python
#Outliers-Age
df["CustomerAge"]=pd.to_numeric(df["CustomerAge"],errors="coerce")
Q1=df["CustomerAge"].quantile(0.25)
Q3=df["CustomerAge"].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
df["CustomerAge"]=df["CustomerAge"].clip(lower,upper)
```

```python
#Outliers-Large Quantities
df["Quantity"]=pd.to_numeric(df["Quantity"],errors="coerce")
Q1=df["Quantity"].quantile(0.25)
Q3=df["Quantity"].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
df["Quantity"]=df["Quantity"].clip(lower,upper)
```

```python
# Missing values
df["CustomerName"]=df["CustomerName"].fillna("Unknown")
df["City"]=df["City"].fillna("Unknown")
df["Product"]=df["Product"].fillna(df["Product"].mode()[0])
df["OrderStatus"]=df["OrderStatus"].fillna(df["OrderStatus"].mode()[0])

df["Quantity"]=df["Quantity"].fillna(df["Quantity"].median())
df["UnitPrice"]=df["UnitPrice"].fillna(df["UnitPrice"].median())
df["CustomerAge"]=df["CustomerAge"].fillna(df["CustomerAge"].median())

df["OrderDate"]=df["OrderDate"].fillna(df["OrderDate"].mode()[0])
```

```python
#Create Total Amount
df["Quantity"]=pd.to_numeric(df["Quantity"],errors="coerce")
df["Total_Amount"]=df["Quantity"]*df["UnitPrice"]
```
